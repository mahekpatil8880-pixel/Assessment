Q1. Data Science Lifecycle

-Problem Definition: The goal is to predict which restaurant zones will have peak demand between 7–9 PM.
-Data Collection: Collect two years of order data, including order time, restaurant zone, and order volume.
-Data Preparation: Clean missing values, duplicates, incorrect entries, and inconsistent data formats.
-Exploration: Analyse demand patterns by zone, day, and time using charts and summary statistics.
-Modelling: Train a suitable machine learning model using historical order data to predict peak demand.
-Evaluation: Compare predictions with actual demand and measure how accurately the model performs.
-Generative AI can speed up Data Preparation by suggesting cleaning code, but incorrect AI-generated code could modify useful data.
-It can also help with Exploration by suggesting charts and patterns, but its interpretations must be checked by a human.


Q2. Git Branching Strategy

-A Git branching strategy allows each developer to work separately without directly changing the working main branch.
-First, each developer gets the latest code using git pull origin main.
-They create their own branch using git checkout -b feature/data-cleaning.
-After completing their work, they use git add . and git commit to save their changes.
-They push the branch using git push -u origin feature/data-cleaning.
-The developer then creates a Pull Request from the feature branch to the main branch for review.
-A merge conflict can occur if two developers edit the same lines of the same file.
-The conflict is resolved manually, the code is tested, and the corrected changes are committed before completing the merge.


Q3. map(), lambda, filter() and reduce()

-First, create a named function that contains the commission calculation logic for each order.
-Use map() with a lambda to apply this function to every order in the list.
-map() is suitable because the same calculation needs to be performed on every record.
-It avoids writing another manual for loop for the same transformation and makes the operation clear.
-Use filter() when the finance team needs only records whose commission or amount is above a given threshold.
-filter() keeps only the records that satisfy the specified condition.
-Use functools.reduce() to combine all calculated commission values and produce one overall total.
-Therefore, map() transforms data, filter() selects data, and reduce() combines data into one result.


Q4. Inheritance and Method Overriding

-Create a base User class containing the common properties and behaviour of all users.
-Customer, DeliveryPartner, and Admin can inherit from this base class.
-Each subclass can override display_profile() to show information specific to that type of user.
-This makes the code organised and avoids putting all different profile logic into one class.
-If display_profile() exists only in some subclasses, a mixed list may cause an AttributeError when the method is missing.
-The solution is to define display_profile() in the base User class so every user object has the method.
-Subclasses can then override the base method with their own profile display.
-This design allows a loop to safely call display_profile() even when new subclasses are added later.


Q5. API Status Codes and Exception Handling

-Status code 200 means the API request was successful, so the program can process the response normally.
-Status code 401 means the request is unauthorized, usually because authentication details are invalid or missing.
-Status code 429 means too many requests were made, so the program should respect the rate limit and retry appropriately.
-Status code 500 means there is a server-side error, so the program should handle the failure instead of processing invalid data.
-The file-writing operation should be placed inside a try block so errors can be caught safely.
-The except block can handle the exception and prevent the program from crashing unexpectedly.
-The finally block always executes, whether an exception occurs or not, so cleanup or completion actions can be performed.
-Proper exception handling helps prevent incomplete file operations from leaving the program in an unexpected state.


Q6. Regular Expressions with re

-Python's re module can be used to validate and extract correctly formatted phone numbers from the raw customer data.
-The regular expression should define the expected number of digits and the allowed formatting characters.
-Depending on the required format, the pattern may allow spaces, hyphens, or a country-code prefix.
-It should reject entries containing letters, incorrect digit counts, or other invalid characters.
-re.findall() is appropriate because it extracts all phone numbers that match the pattern from the text.
-re.search() normally finds only the first matching occurrence, so it is less suitable for extracting every valid entry.
-A real-world edge case is a valid number such as +91 98765 43210, which a simple digits-only pattern might reject.
-Therefore, the regex should be designed according to the actual phone-number formatting rules required by the organisation.
