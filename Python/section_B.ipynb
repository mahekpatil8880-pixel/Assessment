```python
#task-2
restaurants = ["  zomato  ", "SWIGGY", "  dominos ", " McDonald's", "  PIZZA HUT","","12345", "  uber eats ","  blinkit","   "]

amounts = ["Rs 150","200.50","Rs 350","450","Rs 250","100","500","Rs 175","300","125"]
```

```python
valid_records = 0
skipped_records = 0
running_total = 0

clean_restaurants = []

for i in range(len(restaurants)):

    name = restaurants[i].strip()

    if name == "" or name.isdigit():
        print("Skipped invalid entry:", restaurants[i])
        skipped_records += 1
        continue

    name = name.lower()
    name = name.replace("  ", " ")
    name = name.title()

    amount = amounts[i].replace("Rs", "").strip()
    amount = float(amount)

    running_total += amount

    clean_restaurants.append((name, amount))
    valid_records += 1

    print("Valid:", name, "-", amount)
```

```python
print("\n----- FINAL SUMMARY -----")
print("Total valid records:", valid_records)
print("Total skipped records:", skipped_records)
print("Running total amount:", running_total)
```

```python
#task-3
orders = [
    {"id": 1, "name": "Zomato", "amount": 150},
    {"id": 2, "name": "Swiggy", "amount": 250},
    {"id": 3, "name": "Dominos", "amount": 400},
    {"id": 4, "name": "Pizza Hut", "amount": 600},
    {"id": 5, "name": "McDonald's", "amount": 800},
    {"id": 6, "name": "Uber Eats", "amount": 300}
]
```

```python
def calculate_commission(amount):
    
    if amount <= 200:
        return amount * 0.10
    
    elif amount <= 500:
        return amount * 0.15
    
    else:
        return amount * 0.20
```

```python
orders = list(
    map(
        lambda order: {
            **order,
            "fee": calculate_commission(order["amount"])
        },
        orders
    )
)

print("All Orders with Commission:")

for order in orders:
    print(order)
```

```python
threshold = 50

audit_records = list(
    filter(
        lambda order: order["fee"] > threshold,
        orders
    )
)

print("\n----- AUDIT RECORDS -----")

for order in audit_records:
    print(order)
```

```python
from functools import reduce

total_fee = reduce(
    lambda total, order: total + order["fee"],
    orders,
    0
)

print("\nTotal Commission Fee:", round(total_fee, 2))
```

```python
#task-4
import json

class Order:

    def __init__(self, order_id, restaurant_name, amount, delivery_time_minutes, is_delivered):
        self.order_id = order_id
        self.restaurant_name = restaurant_name
        self.amount = amount
        self.delivery_time_minutes = delivery_time_minutes
        self.is_delivered = is_delivered

    def to_dict(self):
        return {
            "order_id": self.order_id,
            "restaurant_name": self.restaurant_name,
            "amount": self.amount,
            "delivery_time_minutes": self.delivery_time_minutes,
            "is_delivered": self.is_delivered
        }

    @classmethod
    def from_dict(cls, data):
        return cls(
            data["order_id"],
            data["restaurant_name"],
            data["amount"],
            data["delivery_time_minutes"],
            data["is_delivered"]
        )

records = [
    Order(1, "Zomato", 250.50, 30, True),
    Order(2, "Swiggy", 400.00, 40, True),
    Order(3, "Dominos", 550.75, 45, False),
    Order(4, "Pizza Hut", 300.25, 35, True)
]

for record in records:
    print(record.to_dict())
```

```python
def save_records(records, filepath):

    try:
        with open(filepath, "w") as file:
            data = []

            for record in records:
                data.append(record.to_dict())

            json.dump(data, file, indent=2)

        print("Records saved successfully.")

    except Exception as e:
        print("Error:", e)

    finally:
        print("Operation complete")
```

```python
def load_records(filepath):

    try:
        with open(filepath, "r") as file:
            data = json.load(file)

        records = []

        for item in data:
            record = Order.from_dict(item)
            records.append(record)

        print("Records loaded successfully.")
        return records

    except FileNotFoundError:
        print("File not found.")
        return []

    except Exception as e:
        print("Error:", e)
        return []

    finally:
        print("Operation complete")
```

```python
filepath = "records.json"

save_records(records, filepath)

loaded_records = load_records(filepath)

print("\n----- LOADED RECORDS -----")

for record in loaded_records:
    print(record.to_dict())
```
