```python
import requests
import json


class Order:
    def __init__(self, order_id, name, amount, status):
        self.order_id = order_id
        self.name = name
        self.amount = amount
        self.status = status

    def to_dict(self):
        return {
            "id": self.order_id,
            "name": self.name,
            "amount": self.amount,
            "status": self.status
        }


records = []


def fetch_restaurants():
    url = "https://jsonplaceholder.typicode.com/posts?_limit=10"

    response = requests.get(url)

    if response.status_code == 200:
        data = response.json()

        records.clear()

        for item in data:
            order = Order(
                item["id"],
                item["title"],
                100,
                "Available"
            )
            records.append(order)

        print("Top 10 Restaurants:")

        for record in records:
            print(record.order_id, "-", record.name)

    else:
        print("API Error:", response.status_code)


def add_restaurant():
    order_id = input("Enter restaurant ID: ")
    name = input("Enter restaurant name: ")

    while True:
        try:
            amount = float(input("Enter amount: "))
            break
        except ValueError:
            print("Invalid amount. Please enter a number.")

    status = input("Enter status: ")

    order = Order(order_id, name, amount, status)
    records.append(order)

    print("Restaurant added successfully.")


def calculate_commission(amount):
    if amount <= 200:
        return amount * 0.10
    elif amount <= 500:
        return amount * 0.15
    else:
        return amount * 0.20


def show_commission():
    commissions = list(
        map(
            lambda record: calculate_commission(record.amount),
            records
        )
    )

    print("\nCommission for all records:")

    for record, commission in zip(records, commissions):
        print(record.name, ":", round(commission, 2))

    filtered_records = list(
        filter(
            lambda record: calculate_commission(record.amount) > 50,
            records
        )
    )

    print("\nRecords with commission above 50:")

    for record in filtered_records:
        print(
            record.name,
            ":",
            round(calculate_commission(record.amount), 2)
        )


def save_and_load():
    filepath = "data/processed/records.json"

    data = [record.to_dict() for record in records]

    with open(filepath, "w") as file:
        json.dump(data, file, indent=2)

    print("Records saved successfully.")

    with open(filepath, "r") as file:
        loaded_data = json.load(file)

    records.clear()

    for item in loaded_data:
        order = Order(
            item["id"],
            item["name"],
            item["amount"],
            item["status"]
        )
        records.append(order)

    print("Records loaded successfully.")


while True:

    print("\nFood Delivery Live Data Pipeline")
    print("1. Fetch and display top 10 restaurants")
    print("2. Add a new restaurant")
    print("3. Calculate commission")
    print("4. Save and load records")

    choice = input("Enter your choice: ")

    if choice == "1":
        fetch_restaurants()

    elif choice == "2":
        add_restaurant()

    elif choice == "3":
        show_commission()

    elif choice == "4":
        save_and_load()

    else:
        print("Invalid choice. Please enter 1, 2, 3 or 4.")
```
