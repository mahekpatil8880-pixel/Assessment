```python
#prompt

"""Write a Python program for a Food Delivery platform that:

1. Fetches the first 5 records from https://jsonplaceholder.typicode.com/posts?_limit=5 and treats them as mock restaurants. Print each record's id and title.

2. Applies a tiered commission to a hardcoded list of 6 order amounts:
   - 10% for orders up to Rs 200
   - 15% for orders from Rs 201 to Rs 500
   - 20% for orders above Rs 500

Use map() and lambda for the commission calculation and print each original amount with its calculated commission.

3. Handles API failure. If the status code is not 200, print a descriptive error message and stop execution without crashing.

4. Saves the 5 fetched records to data/processed/records.json using json.dump() inside a with open() context manager, with indent=2.

Use Python and the requests module."""
```

```python
# AI code

import requests
import json

url = "https://jsonplaceholder.typicode.com/posts?_limit=5"

response = requests.get(url)

if response.status_code == 200:
    restaurants = response.json()

    print("Top 5 Mock Restaurants:")

    for restaurant in restaurants:
        print("ID:", restaurant["id"])
        print("Name:", restaurant["title"])
        print()
else:
    print("API Error:", response.status_code)

amounts = [150, 250, 400, 600, 800, 200]

def calculate_commission(amount):
    if amount <= 200:
        return amount * 0.10
    elif amount <= 500:
        return amount * 0.15
    else:
        return amount * 0.20

commissions = list(
    map(lambda amount: calculate_commission(amount), amounts)
)

print("Commission Results:")

for i in range(len(amounts)):
    print(
        "Original Amount:", amounts[i],
        "Commission:", commissions[i]
    )

with open("data/processed/records.json", "w") as file:
    json.dump(restaurants, file, indent=2)

print("Records saved successfully.")
```

```python
#Corrected version

import requests
import json

url = "https://jsonplaceholder.typicode.com/posts?_limit=5"

try:
    response = requests.get(url)

    if response.status_code == 200:
        restaurants = response.json()

        print("Top 5 Mock Restaurants:")

        for restaurant in restaurants:
            print("ID:", restaurant["id"])
            print("Name:", restaurant["title"])
            print()

        amounts = [150, 250, 400, 600, 800, 200]

        def calculate_commission(amount):
            if amount <= 200:
                return amount * 0.10
            elif amount <= 500:
                return amount * 0.15
            else:
                return amount * 0.20

        commissions = list(
            map(lambda amount: calculate_commission(amount), amounts)
        )

        print("Commission Results:")

        for i in range(len(amounts)):
            print(
                "Original Amount:", amounts[i],
                "Commission:", commissions[i]
            )

        with open("data/processed/records.json", "w") as file:
            json.dump(restaurants, file, indent=2)

        print("\nRecords saved successfully.")

    else:
        print("API Error:", response.status_code)

except Exception as e:
    print("Something went wrong:", e)
```

Explanation:-

The AI version did not handle unexpected request errors such as connection failure.
This could cause the program to stop with an unhandled exception.
I added a try-except block to handle unexpected errors safely.
I also kept the status_code check before calling response.json().
